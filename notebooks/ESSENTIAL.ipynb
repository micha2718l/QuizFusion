```python
from testifyr import (
    WorkProblem,
    Question,
    Test,
    MultipleChoiceProblem,
    Choice,
    Bonus,
    Information,
)
from stuff import Value, Unit
import numpy as np
import sympy as sp
from pprint import pprint as pp
import base64

problems = []
multiple_choices = []
bonuses = []
g = Value(9.81, "m/s^2")
mode = "dev"
```

```python
statement = "Solve the following equations for x."
x, y = sp.symbols("x y")
eq1 = sp.Eq(x**2 - 2, 5)
eq2 = sp.Eq(x**2 - 2, 4)

questions = [
    Question(f'${sp.latex(eq1)}$'),
    Question(f'${sp.latex(eq2)}$'),
]
p = WorkProblem(statement, questions=questions)
problems.append(p)
if mode == "dev":
    p.show()
```
